```python
import matplotlib.pyplot as plt


dna='ATGCCTGAATTCGGCTTAAGCATTCTACGACTC'
count_A=dna.count('A')
count_T=dna.count('T')
count_G=dna.count('G')
count_C=dna.count('C')
print(count_A,count_T,count_G,count_C)
#
gc=(count_G+count_C)/len(dna)*100
print(gc)

counts={'A':0,'T':0,'G':0,'C':0}
for base in dna:
    counts[base]+=1
print(counts)
```
